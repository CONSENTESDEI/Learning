```python
from shlex import join

"""
编写一个函数来查找字符串数组中的最长公共前缀。

如果不存在公共前缀，返回空字符串 ""。



示例 1：

输入：strs = ["flower","flow","flight"]
输出："fl"

示例 2：

输入：strs = ["dog","racecar","car"]
输出：""
解释：输入不存在公共前缀。



提示：

    1 <= strs.length <= 200
    0 <= strs[i].length <= 200
    strs[i] 如果非空，则仅由小写英文字母组成
"""
```

```python
# 解法1
"""
横向扫描
"""
class Solution(object):
    def longestCommonPrefix(self, strs):
        """
        :param strs: List[str]
        :return: str
        """
        # 存储第一个字符串的字符为列表
        prefix = list(strs[0])

        # 匹配找最长公共字符串
        for s in strs[1 : ]: # 遍历每个字符串
            i = 0
            while i < len(s) and i < len(prefix) and s[i] == prefix[i]:
                i += 1
            prefix = prefix[:i] # 删减匹配列表
            if not prefix:
                return ""   # 匹配列表为空，返回空串
        return "".join(prefix)  # 字符串添加新字符

```

```python
# 解法2
"""
横向扫描，排序+首尾比较优化
ASCII保证字母大小顺序固定，a-z
字符串排序后，开头字符小的一定在前，开头字符大的一定靠后
最小字符和最大字符差异最大，若二者存在公共串，则为全部字符串的公共串
"""
class Solution(object):
    def longestCommonPrefix(self, strs):
        """
        :param strs: List[str]
        :return: str
        """
        str_ret = ""
        strs.sort()
        i = 0
        while i < len(strs[0]) and i < len(strs[-1]):
            if strs[0][i] == strs[-1][i]:
                str_ret += (strs[0][i])
                i += 1
            else:
                break
        return str_ret
```

```python
# 解法2
"""
横向扫描，排序+首尾比较优化
减少辅助字符串
"""
class Solution(object):
    def longestCommonPrefix(self, strs):
        """
        :param strs: List[str]
        :return: str
        """
        strs.sort()
        i = 0
        while i < len(strs[0]) and i < len(strs[-1]):
            if strs[0][i] != strs[-1][i]:
                break
            i += 1
        return strs[0][ : i]
```
