```python
"""给你一个整数 x ，如果 x 是一个回文整数，返回 true ；否则，返回 false 。

是指正序（从左向右）和倒序（从右向左）读都是一样的整数。

    例如，121 是回文，而 123 不是。



示例 1：

输入：x = 121
输出：true

示例 2：

输入：x = -121
输出：false
解释：从左向右读, 为 -121 。 从右向左读, 为 121- 。因此它不是一个回文数。

示例 3：

输入：x = 10
输出：false
解释：从右向左读, 为 01 。因此它不是一个回文数。



提示：

    -231 <= x <= 231 - 1



进阶：你能不将整数转为字符串来解决这个问题吗？"""
```

```python
# 解法1
"""
存所有数字到列表
"""
class Solution(object):
    def isPalindrome(self, x):
        """
        :param x: int
        :return: bool
        """
        list = []
        if x < 0:
            return False
        if x < 10:
            return True
        temp = x
        while temp > 0:
            list.append(temp % 10)
            temp = temp // 10
        for i in range(len(list) // 2):
            if list[i] != list[-i - 1]:
                return False
        return True
```

```python
# 解法2
"""
存后半数字作为新数
"""
class Solution(object):
    def isPalindrome(self, x):
        """
        :param x: int
        :return: bool
        """
        if x < 0 or x % 10 == 0:
            return False
        if x < 10:
            return True
        temp = 0
        while x > temp:
            temp = temp * 10 + x % 10
            x //= 10
        if x == temp or x == temp // 10:
            return True
        return False
```

```python
# 解法3
"""
转换为字符串
"""
class Solution(object):
    def isPalindrome(self, x):
        """
        :param x: int
        :return: bool
        """
        if x < 0:
            return False
        temp = str(x)
        for i in range(len(temp) // 2):
            if temp[i] != temp[-i - 1]:
                return False
        return True
```
