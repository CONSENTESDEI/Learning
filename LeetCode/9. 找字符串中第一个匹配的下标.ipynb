```python
"""
给你两个字符串 haystack 和 needle ，请你在 haystack 字符串中找出 needle 字符串的第一个匹配项的下标（下标从 0 开始）。如果 needle 不是 haystack 的一部分，则返回  -1 。



示例 1：

输入：haystack = "sadbutsad", needle = "sad"
输出：0
解释："sad" 在下标 0 和 6 处匹配。
第一个匹配项的下标是 0 ，所以返回 0 。

示例 2：

输入：haystack = "leetcode", needle = "leeto"
输出：-1
解释："leeto" 没有在 "leetcode" 中出现，所以返回 -1 。



提示：

    1 <= haystack.length, needle.length <= 104
    haystack 和 needle 仅由小写英文字符组成
"""
```

```python
# 解法1
"""
暴力匹配
让 pre 指针遍历全部字符串
"""
class Solution(object):
    def strStr(self, haystack: str, needle: str) -> int:
        if len(haystack) < len(needle):
            return -1
        pre = 0
        rear = 0
        while pre != len(haystack):
            rear = pre
            i = 0
            while i != len(needle) and rear < len(haystack):
                if haystack[rear] != needle[i]:
                    break
                else:
                    i += 1
                    rear += 1
            if i == len(needle):
                return pre
            else:
                pre += 1
        return -1



```

```python
# 解法2
"""
KMP跳转数组
"""
"""
针对next数组的理解
1. maxPre：是最大前后缀长度，也是 前缀 的下一个需要比较 是否为前缀的 字符，
            所以他可以作为该字符的索引，用于和 i 所指字符 进行比较
2. 回溯：当lis[maxPre] != lis[i] 时
        maxPre = next[maxPre - 1] 是为了确认是否存在短一点的最长公共前后缀，
        后续以新的 maxPre 作为下标和 i 指向元素比较，
        直到MaxPre == 0，证明当前遍历过的字符串不存在包含 i 所指元素的最长公共前后缀，那么他的next就是0
3. 对比部分：因为maxPre指向的永远是前缀 的下一个字符，是需要判定是否要加入最长前缀末尾，
            所以要和 lis[i] 比较，因为 i 之前的全部确定过了。
"""
class Solution(object):
    # 寻找next数组
    @staticmethod
    def find_next(lis):
        next = [0]
        maxPre = 0
        i = 1
        while i < len(lis):
            if lis[maxPre] == lis[i]:
                maxPre += 1
                next.append(maxPre)
                i += 1
            else:
                if maxPre == 0:
                    next.append(maxPre)
                    i += 1
                else:
                    maxPre = next[maxPre - 1]
        return next

    # 寻找第一个相同部分的下标
    def strStr(self, haystack: str, needle: str) -> int:
        if len(haystack) < len(needle):
            return -1

        next = Solution.find_next(needle)

        h_len = len(haystack)
        n_len = len(needle)
        i = 0
        j = 0

        while i < h_len:
            if haystack[i] == needle[j]:
                i += 1
                j += 1
            else:
                if j > 0:
                    j = next[j - 1]
                else:
                    i += 1
            if j == n_len:
                return i - j
        return -1
```

```python
# 解法3
"""
切片比较
"""
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        m,n = len(needle),len(haystack)

        for i in range(n - m +1):
            if haystack[i:i+m] == needle:
                return i
        return -1
```
