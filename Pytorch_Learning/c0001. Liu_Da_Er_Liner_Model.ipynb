```python
# 导包
import numpy as np
import matplotlib.pyplot as plt

# 1.导入数据
x_data = [1.0, 2.0, 3.0]
y_data = [2.0, 4.0, 6.0]

# 2.确定模型或对应关系
def forward(x):
    return x * w

# 3.损失函数
def loss(x, y):
    y_pred = forward(x)
    return (y_pred - y) ** 2

# 4.创建权重表和 MSE表
w_list = []
mes_lest = []

# 5.生成 权重w 算出损失函数和MSE值
for w in np.arange(0.0, 4.1, 0.1):
    print(f"w={w:2f}")
    l_sum = 0
    for x_val, y_val in zip(x_data, y_data):
        y_pred_val = forward(x_val)
        loss_val = loss(x_val, y_val)
        l_sum += loss_val
        print(f"\t {x_val:.1f} {y_val:.1f} {y_pred_val:.3f} {loss_val:.4f}")
    print("MSE = ", l_sum / len(x_data))
    w_list.append(w)
    mes_lest.append(l_sum / len(x_data))

# 6.输出权重和MSE的统计图
plt.plot(w_list, mes_lest)
plt.ylabel("loss")
plt.xlabel("w")
plt.savefig('./resource/picture/loss_plot.png')
plt.show()
```

```python
"""
复现
"""
# 导包
import numpy as np
import matplotlib.pyplot as plt

# 1.导入数据
x_data = [1.0, 2.0, 3.0]
y_data = [2.0, 4.0, 6.0]
len_x = len(x_data)
# 2.设置权重函数
def forward(x):
    return x * w

# 3.损失函数
def loss(x, y):
    y_pred = forward(x)
    return (y_pred - y) ** 2

# 4.权重表和MSE表
w_list = []
mes_list = []

# 5.不同权重下，各返回值
for w in np.arange(0.0, 4.1, 0.1):
    print(f"w={w:.2f}")
    l_sum = 0
    for x_val, y_val in zip(x_data, y_data):
        # 预测值y
        y_pred_val = forward(x_val)
        # 损失值
        loss_val = loss(x_val, y_val)
        l_sum += loss_val
        print(f"\t {x_val:.1f} {y_val:.1f} {y_pred_val:.3f} {loss_val:.4f}")
    mes_val = l_sum / len_x
    print(f"MSE = {mes_val:.3f}")
    mes_list.append(mes_val)
    w_list.append(w)

# 6.输出图像
plt.plot(w_list, mes_list)
plt.ylabel("MSE")
plt.xlabel("w")
plt.show()
```

```python
"""
作业，加入 b画出(MES, w, b)
"""
# 导包
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1.导入数据
x_data = [1.0, 2.5, 3.2, 4.8]
y_data = [2.5, 5.0, 6.9, 10.1]

# 2.权重函数
def w_d_func(x):
    return x * w + b

# 求出表长
len_x = len(x_data)

# 4.损失函数
def loss(x, y):
    y_pred = w_d_func(x)
    return y_pred, (y_pred - y) ** 2

# 4.w b mse 的列表
w_list = []
b_list = []
mse_list = []

# 5.不同权重下的各返回值
for w in np.arange(0.0, 5.1, 0.1):
    print(f"w={w:.1f}")
    for b in np.arange(-1.0, 1.1, 0.1):
        loss_sum = 0
        print(f"\tb={b:.1f}")
        for x_val, y_val in zip(x_data, y_data):
            y_pred_val, loss_val = loss(x_val, y_val)
            loss_sum += loss_val
            print(f"\t\t {x_val:.1f}  {y_val:.1f}  {y_pred_val:.3f}  {loss_val:.4f}")
        # 计算mse
        mse_loss = loss_sum / len_x
        print(f"\tMSE = {mse_loss:.3f}")

        # 存入mse
        mse_list.append(mse_loss)
        # 存入b
        b_list.append(b)
    # 存入w
    w_list.append(w)

# 6.绘制图表
# 将列表转换为二维网格
w_unique = np.arange(0.0, 5.1, 0.1)
b_unique = np.arange(-1.0, 1.1, 0.1)
w_grid, b_grid = np.meshgrid(w_unique, b_unique)
cost_grid = np.array(mse_list).reshape(len(b_unique), len(w_unique))

# 然后正常绘图
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(w_grid, b_grid, cost_grid, cmap='viridis', alpha=0.8, edgecolor='none')

# 修改颜色条的位置和大小
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10,
             label='MSE',
             pad=0.2)  # pad控制与主图的间距，增大可以避免遮挡

ax.set_xlabel("w")
ax.set_ylabel("b")
ax.set_zlabel("MSE")



plt.savefig('./resource/picture/loss_surface.png')
plt.show()
```
