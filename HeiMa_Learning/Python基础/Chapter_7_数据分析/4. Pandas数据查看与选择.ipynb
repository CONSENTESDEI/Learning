```python
df2 = pd.read_csv("./Project_Data/4. Pandas_sales_data_2.csv" , encoding = 'UTF-8')
df2
```

```python
# 单项选择
df2[df2['销售数量'] >= 10]
```

```python
# 单项复合选择
# df2[(df2['销售数量'] >= 10) & (df2['销售数量'] <= 13) & (df2['单价'] <= 1000)]

# 用 between 简化
df2[(df2['销售数量'].between(10, 13)) & (df2['单价'] <= 1000)]
```

```python
df2[
    ((df2['产品类别'].isin(['食品饮料'])) & (df2['销售数量'].between(5, 10, inclusive= 'neither'))) |
    ((df2['产品类别'].isin(['办公耗材'])) & (df2['销售数量'].between(10,15, inclusive= 'left')))
]
```
