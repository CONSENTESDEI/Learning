```python
"""
KNN 鸢尾花案例，寻找最优超参

网格搜索和交叉验证

交叉验证：
    略

网格搜索：
    略

学习的对象是 GribSearchCV
"""
```

```python
# 导包

# 加载鸢尾花数据集（内置经典多分类数据集，含150个样本，3个类别）
from sklearn.datasets import load_iris

# 数据集划分工具，将数据按指定比例随机分割为训练集和测试集
from sklearn.model_selection import train_test_split, GridSearchCV

# 数据标准化处理器，将特征缩放为均值为0、方差为1的标准正态分布（提升KNN等距离算法的性能）
from sklearn.preprocessing import StandardScaler

# K近邻分类器（K-Nearest Neighbors），基于距离度量（如欧氏距离）进行分类的监督学习算法
from sklearn.neighbors import KNeighborsClassifier

# 模型评估指标，计算分类正确的样本数占总样本数的比例（准确率）
from sklearn.metrics import accuracy_score
```

```python
# 1.加载数据集
iris_data = load_iris()

# 2.数据预处理，8:2
x_train, x_test, y_train, y_test = train_test_split(iris_data.data,
                                                    iris_data.target,
                                                    test_size = 0.2,
                                                    random_state = 20)

# 3.特征工程，标准化
# 3.1.创建标准化对象
transfer = StandardScaler()

# 3.2.对数据进行标准化
x_train = transfer.fit_transform(x_train)
x_test = transfer.transform(x_test)

# 4.模型训练
# 4.1.创建KNN分类对象
estimator = KNeighborsClassifier()

# 4.2.定义字典，记录超参可能出现的情况
param_dict = {"n_neighbors": range(1, 11)}

# 4.3.创建 GridSearchCV 寻找最优超参，用网格搜索和交叉验证
"""
参数1：计算最优超参的模型对象
参数2：模型超参可能出现的值
参数3：交叉验证折数
返回值：处理后的模型对象
"""
estimator = GridSearchCV(estimator, param_dict, cv=4)

# 4.5.模型训练
estimator.fit(x_train, y_train)

# 4.5.打印最优超参组合
print((f"最优评分：{estimator.best_score_}\n"
       f"最优超参组合：{estimator.best_params_}\n"
       f"最优估计器对象：{estimator.best_estimator_}\n"
       f"具体的每轮验证结果：{estimator.cv_results_}"))

# 5.模型评估
# 5.1.获取最优超参的模型对象
estimator = estimator.best_estimator_

# 5.2.模型训练
estimator.fit(x_train, y_train)

# 5.3.模型预测
y_pred = estimator.predict(x_test)

# 5.4.模型评估
print(f"准确率：{accuracy_score(y_test, y_pred)}")
```
